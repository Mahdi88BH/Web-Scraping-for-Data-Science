## **What is Packet Switching?**

- **Packet switching** is the fundamental method of transmitting data across modern computer networks (including the internet).

> Instead of opening a single, continuous dedicated electrical or optical connection between a sender and receiver (like traditional telephone lines, known as *circuit switching*), packet switching breaks down data into small, manageable chunks called **packets**. Each packet travels independently across the network and is reassembled at the destination.

---

### **Core Concepts & How It Works**

To understand packet switching, imagine sending a massive 500-page book to a friend. Instead of renting a dedicated truck to drive the entire book across the country all at once (circuit switching), you tear out individual pages, put each page in its own envelope, mail them separately through various postal routes, and have your friend reassemble the book at the other end.

### **1. Packet Structure (The Envelope)**

- Every packet consists of two primary components:
    * **Header:** Contains metadata such as the **source IP address**, **destination IP address**, **packet sequence number**, and error-checking data. Routers use this header to figure out where to forward the packet.
    * **Payload:** The actual chunk of user data being transmitted (e.g., a fragment of a web page, an image, or a video frame).

### **2. Independent Routing**

As packets leave your computer, they enter a network of routers. Each router inspects the destination header of an incoming packet and decides the most efficient next hop based on current network traffic.

* **Packet A** might take Route 1 (via fiber optic cable).
* **Packet B** might take Route 2 (via satellite link) because Route 1 experienced congestion.

### **3. Reassembly & Sequencing**

Because packets take different paths through the network, they often arrive out of order (or occasionally get lost and need to be retransmitted). Protocols like **TCP (Transmission Control Protocol)** use the sequence numbers in the packet headers to place them back into the exact right order before presenting the data to your application.

---

### **The Two Main Types of Packet Switching**

| Type | How it Works | Use Case |
| --- | --- | --- |
| **Datagram Packet Switching (Connectionless)** | Every packet is treated independently. Different packets in the same file can take completely different paths. | **The Internet (IP packets)**. Highly resilient because if a router fails, subsequent packets automatically route around it. |
| **Virtual Circuit Packet Switching (Connection-Oriented)** | A pre-planned logical route (a "virtual path") is established before any packets are sent. All packets follow this exact path. | Frame Relay, MPLS networks. Offers more predictable delivery times and ordered arrival. |

---

### **Why Packet Switching Won Over Circuit Switching**

* **Network Efficiency:** Circuit switching locks up a dedicated circuit for the entire duration of a session—even if nobody is talking or downloading anything (wasting bandwidth). Packet switching shares network links dynamically among thousands of simultaneous users.
* **Fault Tolerance:** If a physical cable or router breaks along a route in a packet-switched network, traffic instantly reroutes through alternative paths. In circuit switching, the call or connection drops immediately.
* **Cost-Effective Scaling:** Building infrastructure for packet networks is significantly cheaper and scales seamlessly from a local home network to global internet backbones.
